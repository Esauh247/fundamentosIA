```python
##Grafos dirigidos
import matplotlib.pyplot as plt 
import networkx as nx
```

```python
#Creamos el grafo
grafo = nx.DiGraph()
grafo.add_weighted_edges_from(
    [
        ("A","B",5),
        ("A","C",2),
        ("B","D",4),
        ("C","E",5),
        ("C","F",6)
    ]
)
```

```python
#Imprimimos el grafo
pos = nx.spring_layout(grafo,seed=10)
nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
    
)
labels = nx.get_edge_attributes(grafo,"weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels = labels
)
plt.show()
```

```python
#¿puedo moverme de A a B?
print (grafo.has_edge("A","B"))
```

```python
#puedo moverme de B a A?
print(grafo.has_edge("B","A"))
```

```python
#puedo moverme de A a F?
print(grafo.has_edge("A","F"))
```

```python
#Mostramos las rutas
rutas=list (
    nx.all_simple_paths(
        grafo,
        source="A",
        target="F"
    )
)
```

```python
#imprimimos las rutas
for i,ruta in enumerate(rutas,start = 1):
    print(f"Ruta {i}:{ruta}")
```

```python
#Calculamos el costo
def costo_ruta(grafo2,ruta):
    costo=0
    for i in range (len(ruta)-1):
        costo += grafo2[ruta[i]][ruta[i+1]]["weight"]
        return costo
```

```python
#Imprimimos el costo
for i, ruta in  enumerate(rutas,start=1):
    costo=costo_ruta(grafo,ruta)
    print(f"Ruta{i}:{ruta}->{costo} unidades")
```
