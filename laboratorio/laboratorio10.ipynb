```python
##Grafos dirigidos
import matplotlib.pyplot as plt 
import networkx as nx
```

```python
#Creamos el grafo
grafo = nx.DiGraph()
grafo.add_weighted_edges_from(
    [
        ("centro","sur",15),
        ("centro","norte",10),
        ("centro","este",25),
        ("norte","sur",30),
        ("norte","este",8),
        ("norte","oeste",15),
        ("sur","oeste",9),
        ("este","aeropuerto",20),
        ("oeste","aeropuerto",18)
        
    ]
)
```

```python
#Imprimimos el grafo
pos = nx.spring_layout(grafo,seed=11)
nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
    
)
labels = nx.get_edge_attributes(grafo,"weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels = labels
)
plt.show()
```

```python
#mostrar rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source = "centro",
        target = "aeropuerto"
    )
)
#imprimir rutas
for i, ruta in enumerate (rutas, start=1):
    print (f"Ruta{i}:{ruta}")
    
```

```python
#Utilizamos algoritmos para obtener la ruta corta
ruta_mas_corta = nx.dijkstra_path(grafo,source="centro",target="aeropuerto",weight="weight")
tiempo_total=nx.dijkstra_path_length(grafo,source="centro",target="aeropuerto",weight="weight")
```

```python
#Imprimir
print(f"La ruta más corta es:{ruta_mas_corta}")
print(f"El tiempo total es:{tiempo_total} minutos")
```
