```python
import matplotlib.pyplot as plt ##Librerias para graficas 
import networkx as nx ##libreria que me permite trabajar con estructuaras complejas, como grafos
```

```python
##creamos una instacia de la clase graph
## graph es una coleccion de nodos
grafo = nx.Graph()
```

```python
## creamos el grafo utilizanfo el metodo add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("A","E"),
        ("B","D"),
        ("D","F"),
        ("E","G"),
        ("E","F")
    ]
)
```

```python
##Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size=1000,
    node_color="#e44b8d"

)
plt.show()
```

```python
##Mostrar las rutas posibles del grafo 
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="G")
)
```

```python
##Imprimir las rutas
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
##mostrar el total de pasos de cada ruta 
pasos1 = len(rutas[0])-1
print("pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("pasos de la ruta 2: ", pasos2)
```

```python
##Aplicamos la logica proposicional
p = pasos1 < pasos2 #p: la mejor ruta es la 1 
Q = pasos2 < pasos1 #La mejor ruta es la 2 

if p:
    print("P = True")
    print("La ruta 1 es mejor opcion")
elif Q:
    print("Q = True")
    print("La ruta 2 es mejot opcion")
else:
    print("P= False")
    print("Q=False")
    print("Las dos rutas son igual de buenas")
```
