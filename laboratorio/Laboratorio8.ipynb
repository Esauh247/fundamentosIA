```python
import matplotlib.pyplot as plt ##Librerias para graficas 
import networkx as nx ##libreria que me permite trabajar con estructuaras complejas, como grafos
```

```python
grafo = nx.Graph()
grafo.add_edges_from(
    [
        ("PC1","SWITCHA"),
        ("SWITCHA","IMPRESORAA"),
        ("SWITCHA","ROUTERA"),
        ("ROUTERA","SWITCHB"),
        ("IMPRESORAA","SWITCHB"),
        ("ROUTERA","SERVIDOR_1"),
        ("SWITCHB","PC2")
     
    ]
)
```

```python
nx.draw(
    grafo,
    with_labels = True,
    node_size=1000,
    node_color="#e44b8d"

)
plt.show()
```
